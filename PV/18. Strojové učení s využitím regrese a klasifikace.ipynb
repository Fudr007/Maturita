# 18. Strojové učení s využitím regrese a klasifikace

### Princip strojového učení
* Úlohou je na základě tzv. trénovacích dat nalézt matematickou funkci $f$
* Tato funkce následně pro jakýkoliv nový vstup určí odpovídající výstup
* Klademe důraz na generalizaci oproti memorování

### Rozdíl mezi Klasifikací a Regresí
* Klasifikace
    * Výstupem je nějaká kategorie (třída)
    * Využívá se například pro rozlišení jablek a hrušek (podle barvy a tvaru) detekci spamu, strojový překlad jazyků nebo diagnostiku onemocnění
* Regrese
    * Výstupem je číselná hodnota
    * Typickým příkladem je predikce ceny bytu v milionech Kč na základě jeho vlastností, předpověď teploty nebo ceny Bitcoinu

### Lineární a Logistická regrese
* Lineární regrese (pro regresi) - hledá parametry (váhy $w$), kterými se vynásobí vstupy, cílem trénování je minimalizovat celkovou chybu, která se měří jako MSE (kvadratická chyba) nebo MAE (absolutní chyba)
*  Logistická regrese (pro klasifikaci) - namísto MSE minimalizuje při trénování křížovou entropii ($e$), pro hodnocení její kvality se poté používají metriky jako Accuracy, Precision, Recall a F-míra, může fungovat binárně, nebo pro více tříd strategiemi "jeden versus ostatní" či "jeden versus jeden"

### Random forest a decision tree
* Decision tree - pro regresi i klasifikaci, trénováním vzniká binární strom, v uzlech jsou podmínky a v koncových listech je uložena výsledná predikce
* Random forest - modely složené z více decision tree, u regrese výsledek stromů zprůměrují, u klasifikace predikují na základě většinového "hlasování"

### Přetrénování (Overfitting)
* Model si zapamatuje trénovací data, ale ztratí schopnost zobecňovat na data nová
* Chyba u testovacího data setu je výrazně vyšší než chyba u trénovacího data setu
* Prevencí je použití větších datových sad nebo snížení složitosti modelu (menší hloubka)

```python
import numpy as np
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import mean_absolute_error, accuracy_score
```

**1. REGRESE: Predikce ceny bytu**

```python
# Vstupy (Features): [plocha v m2, patro, pocet_mistnosti]
X_reg = np.array([[42, 7, 2], [105, 3, 3], [67, 1, 2], [224, 3, 4]])
# Výstupy (Labels): Cena v mil. Kč
y_reg = np.array([3.2, 6.8, 4.1, 13.9])

model_reg = LinearRegression()
# Natrénování modelu
model_reg.fit(X_reg, y_reg)

# Predikce ceny pro nový neznámý byt: 80m2, 2. patro, 3 místnosti
cena_predikce = model_reg.predict([[80, 2, 3]])
print(f"Predikovaná cena bytu (Regrese): {cena_predikce[0]:.2f} mil. Kč")
```

**2. KLASIFIKACE: Rozlišení jablek a hrušek**

```python
# Vstupy: [hmotnost_v_gramech, barva_kod (0=cervena, 1=zluta, 2=zelena)]
X_clf = np.array([[146, 0], [120, 1], [187, 2], [155, 0]])
# Výstupy (Třídy): 0 = jablko, 1 = hruška
y_clf = np.array([0, 1, 1, 0])

# Vytvoření a natrénování klasifikačního modelu
model_clf = LogisticRegression()
model_clf.fit(X_clf, y_clf)

# Predikce pro neznámý plod (130g, žlutá barva = 1)
druh_predikce = model_clf.predict([[130, 1]])

vysledek_text = 'hruška' if druh_predikce[0] == 1 else 'jablko'
print(f"Predikovaný druh ovoce (Klasifikace): {vysledek_text}")
```
