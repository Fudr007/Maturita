# 24. Výjimky a aserce, debugování a zpracování chyb

### Zpracování chyb a vyjímky
* SW systémy musí být odolné vůči všem chybovým stavům, které při běhu aplikace mohou nastat
* K zachycení a zpracování se využívá konstrukce bloků `try/catch` (v Pythonu zapisováno `try/except`)
* Když v bloku uvnitř `try/catch` nastane chyba, program nekontrolovatelně nespadne, ale provede něco jiného
* Pomocí `catch` lze zachytit různé typy chyb, tedy následně jinak regovat na danou situaci (`ValueError`, `TypeError`, a všem nadřazená `Exception`)
* Nakonec za blok lze přidat `finally` klauzuli, pod tou se provede kód za všech podmínek, ať už program půjde správně dál nebo byl zachycen vyjímkou

### Aserce (Assertions)
* Striktní kontrole předpokladů přímo uvnitř zdrojového kódu, pojistka vlastní logiky
* Vývojářský nástroj, nepoužívat v praxy na produkci (často pri bildu smazány)
* Klíčového slovo assert umožňuje např. v metodě vynutit, že vstupní datový typ je výlučně textový řetězec a má přesnou délku. Pokud výraz není pravdivý, program vyhodí výjimku

### Debugování a zpracování chyb
* Hledání proč kód nefunguje tak jak si představujeme
* Print Debugging - amaterský ale rychlý způsob, pomocné `print` v kódu, abychom viděli co se zrovna v kódu děje. Správně se na to používá knihovna `logging`
* Debugger v IDE - nástroje přímo v IDE pro hledání chyb, profesionální přístup.
    * Breakpoint(červený puntík) = vybere řádek kódu, na kterém se kód zastaví
    * Inspekce paměti = když je program zastavený po breakpointu, tak se mohu podívat co je zrovna v různých proměnných uloženo
    * Step Over/Step Into = krokování, řádek po řádku projíždění zastaveného kódu, přičemž přesně vidím co se děje a proč kód dělá to co dělá

**Vyjimky**

```python
def analyza_zakaznika(celkova_utrata, pocet_nakupu):
    print("--- Začátek analýzy ---")

    try:
        prumer = celkova_utrata / pocet_nakupu

    except ZeroDivisionError:
        print("Chyba: Zákazník zatím nemá žádné nákupy. Nelze spočítat průměr.")
    except TypeError:
        print("Chyba: Do systému přišla poškozená data (text místo čísla).")

    else:
        print(f"✅ Úspěch: Průměrná útrata zákazníka je {prumer} Kč.")

    finally:
        print("🔒 Úklid: Uzavírám složku zákazníka...\n")

analyza_zakaznika(5000, 2)
analyza_zakaznika(0, 0)
```

**Aserce (vyvojarsky nastroj)**

```python
def zdaneni_zbozi(cena_bez_dane):
    cena_s_dani = cena_bez_dane * 1.21  # Přidáme 21 % DPH


    assert cena_s_dani >= cena_bez_dane, "FATÁLNÍ CHYBA LOGIKY: Daň zmenšila cenu!"

    return cena_s_dani

print(zdaneni_zbozi(1000))
```
