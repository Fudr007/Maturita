# 24. Výjimky a aserce, debugování a zpracování chyb

### Zpracování chyb a vyjímky
* SW systémy musí být odolné vůči všem chybovým stavům, které při běhu aplikace mohou nastat
* K zachycení a zpracování se využívá konstrukce bloků `try/catch` (v Pythonu zapisováno `try/except`)
* Když v bloku uvnitř `try/catch` nastane chyba, program nekontrolovatelně nespadne, ale provede něco jiného
* Pomocí `catch` lze zachytit různé typy chyb, tedy následně jinak regovat na danou situaci (`ValueError`, `TypeError`, a všem nadřazená `Exception`)
* Nakonec za blok lze přidat `finally` klauzuli, pod tou se provede kód za všech podmínek, ať už program půjde správně dál nebo byl zachycen vyjímkou

### Aserce (Assertions)
* Striktní kontrole předpokladů přímo uvnitř zdrojového kódu
* Klíčového slovo assert umožňuje např. v metodě vynutit, že vstupní datový typ je výlučně textový řetězec a má přesnou délku. Pokud výraz není pravdivý, program vyhodí výjimku
* Jsou ústředním prvkem metodik Unit testování (`assertEqual()`, `assertNotEqual()`, `assertTrue()`)
* Speciální testovací metoda assertRaises() asertuje, zda cílový kód správně vyhodí požadovaný druh výjimky

### Debugování a reportování chyb
* Pokud aserce, unit testy nebo ruční testování odhalí defekt, je třeba jej formálně zdokumentovat v tzv. Bug Reportu
* Pro debugování a opravu obsahuje report přesné kroky k navození chyby (Steps to reproduce), a porovnání očekávaného (Expected Result) se skutečným chováním (Actual Result)
* U chyb se povinně určuje Severity (závažnost) a Priority (naléhavost)

```python
import unittest
```

**1. ASERCE (Kontrola předpokladů a datových typů)**

```python
def rok_narozeni_z_rc(rodne_cislo):
    # Ochranné aserce, které zajistí zastavení programu při nevalidním vstupu
    assert type(rodne_cislo) == str, "Rodné číslo musí být textový řetězec!"
    assert len(rodne_cislo) == 11, "Rodné číslo musí obsahovat přesně 11 znaků!"

    rok_narozeni = int(rodne_cislo[0:2])
    return rok_narozeni + 1900 if rok_narozeni > 21 else rok_narozeni + 2000
```

**2. VÝJIMKY A ZPRACOVÁNÍ CHYB**

```python
def bezpecny_vypocet(hodnota_str):
    # Návrhový vzor s blokem try/catch (v Pythonu try/except) pro bezpečný chod
    try:
        # Tento převod může selhat, pokud řetězec neobsahuje jen číslice
        cislo = int(hodnota_str)
        return 100 / cislo
    except ValueError as chyba_hodnoty:
        # Zpracování chybných znaků [cite: 1220]
        print(f"Zpracována chyba typu ValueError: Nelze převést '{hodnota_str}'.")
        return None # Návrat None místo vyhození kritické výjimky
    except ZeroDivisionError:
        print("Zpracována chyba typu ZeroDivisionError: Pokus o dělení nulou.")
        return None
```

**3. UNIT TEST PRO TESTOVÁNÍ VÝJIMEK**

```python
class TestZpracovaniChyb(unittest.TestCase):
    def test_spravneho_vyhozeni_vyjimky(self):
        # Aserce, která úspěšně projde pouze tehdy, když kód uvnitř způsobí ValueError
        with self.assertRaises(ValueError):
            # Simulujeme špatný uživatelský vstup obsahující písmena ("AHOJ") pro sčítání/výpočty [cite: 1220]
            rok_narozeni_z_rc("AHOJ1234/56")
```

```python
# V Jupyter notebooku se dají unit testy spustit následujícím způsobem:
if __name__ == '__main__':
    # argv=[''] a exit=False zaručí plynulý běh v buňce Notebooku
    unittest.main(argv=['first-arg-is-ignored'], exit=False)

    # Test try/except bloku:
    print("\n--- Spuštění funkce s ošetřenou výjimkou ---")
    vysledek = bezpecny_vypocet("neplatne_cislo")
    print(f"Konečný výsledek bezpečného výpočtu: {vysledek}")
```
