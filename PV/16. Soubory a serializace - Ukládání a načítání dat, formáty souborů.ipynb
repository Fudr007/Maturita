# 16. Soubory a serializace - Ukládání a načítání dat, formáty souborů

### Práce se soubory
* K perzistentnímu uložení dat mimo operační paměť
* Při otevření souboru je nutné specifikovat mód: `r` (čtení), `w` (zápis, přepisuje stávající obsah), `a` (append - připisuje) a s přidáním `b` i binární režimy (např. `wb` pro binární zápis)
* Soubory je nutné po přečtení vždy uzavřít, aby nedošlo k memory leaks nebo poškození
* Využívá se konstrukce kontextového manažeru `with`, garantuje bezpečné zavření souboru i když program uvnitř bloku nečekaně spadne

### Textové datové formáty
* TXT - Surový neformátovaný text, čte se obvykle po řádcích
* CSV - Plochý formát pro tabulková data, řádky jsou záznamy, sloupce jsou oddělené čárkou/středníkem, k parsování knihovna csv nebo pandas
* JSON - Moderní, stromový, člověkem dobře čitelný formát, principu párů `klíč: hodnota`, v Pythonu se mapuje na strukturu slovníků a polí
* XML - Hierarchický formát využívající strukturu počátečních a koncových tagů, dnes na ústupu kvůli JSONu

### Serializace a Deserializace
* Serializace - Proces převedení složitého datového objektu z operační paměti tak aby šel uložit na disk
* Deserializace - Zpětný proces serializace
* V Pythonu vestavěný modul `pickle`, k zápisu binárního souboru `pickle.dump()`, k načtení `pickle.load()`

```python
import json
import pickle
```

**1. PRÁCE S TEXTOVÝM SOUBOREM**

```python
# Kontextový manažer (with) se postará o automatické uzavření souboru
with open('zaznam.txt', 'w', encoding='utf-8') as file:
    file.write("První řádek uloženého textu.\n")

with open('zaznam.txt', 'r', encoding='utf-8') as file:
    obsah_txt = file.read()
    print("Načteno z TXT:", obsah_txt.strip())
```

**2. ULOŽENÍ A NAČTENÍ VE FORMÁTU JSON**

```python
# Slovník reprezentující strukturovaná data
uzivatel_data = {
    "jmeno": "Jan Novák",
    "vek": 30,
    "opravneni": ["admin", "editor"]
}

# Uložení do JSON
with open('data.json', 'w', encoding='utf-8') as json_file:
    # indent=4 zajistí hezké odsazení pro čitelnost
    json.dump(uzivatel_data, json_file, indent=4)

# Načtení z JSON přímo zpět do slovníku (dict)
with open('data.json', 'r', encoding='utf-8') as json_file:
    nacteny_slovnik = json.load(json_file)
    print("Načteno z JSON - První oprávnění:", nacteny_slovnik['opravneni'][0])
```

**3. PICKLE**

```python
class Hrac:
    def __init__(self, jmeno, level):
        self.jmeno = jmeno
        self.level = level

hrac1 = Hrac("Gamer123", 42)

# Serializace (uložení objektu třídy Hrac do binárního souboru)
# Využíváme mód 'wb' (write binary)
with open('savegame.pkl', 'wb') as pickle_file_out:
    pickle.dump(hrac1, pickle_file_out)

# Deserializace (obnova živého objektu z binárního souboru)
# Využíváme mód 'rb' (read binary)
with open('savegame.pkl', 'rb') as pickle_file_in:
    obnoveny_hrac = pickle.load(pickle_file_in)

print(f"Deserializovaný objekt - Hráč: {obnoveny_hrac.jmeno}, Level: {obnoveny_hrac.level}")
```
