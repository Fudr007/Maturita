# 16. Soubory a serializace - Ukládání a načítání dat, formáty souborů

### Práce se soubory
* K perzistentnímu uložení dat mimo operační paměť
* Při otevření souboru je nutné specifikovat mód: `r` (čtení), `w` (zápis, přepisuje stávající obsah), `a` (append - připisuje) a s přidáním `b` i binární režimy (např. `wb` pro binární zápis)
* Soubory je nutné po přečtení vždy uzavřít, aby nedošlo k memory leaks nebo poškození
* Využívá se konstrukce kontextového manažeru `with`, garantuje bezpečné zavření souboru i když program uvnitř bloku nečekaně spadne

### Textové datové formáty
* TXT - Surový neformátovaný text, čte se obvykle po řádcích
* CSV - Plochý formát pro tabulková data, řádky jsou záznamy, sloupce jsou oddělené čárkou/středníkem, k parsování knihovna csv nebo pandas
* JSON - Moderní, stromový, člověkem dobře čitelný formát, principu párů `klíč: hodnota`, v Pythonu se mapuje na strukturu slovníků a polí
* XML - Hierarchický formát využívající strukturu počátečních a koncových tagů, dnes na ústupu kvůli JSONu

### Serializace a Deserializace
* Serializace - Proces převedení složitého datového objektu z operační paměti tak aby šel uložit na disk
* Deserializace - Zpětný proces serializace
* V Pythonu vestavěný modul `pickle`, k zápisu binárního souboru `pickle.dump()`, k načtení `pickle.load()`

**1. PRÁCE SE SOUBOREM**

```python
import json
seznam_studentu = [
    {"jmeno": "Jan Novak", "avg_znamky":1.7},
    {"jmeno": "Vladimir Hrusensky", "avg_znamky":2.1}
]

with open("seznam.json", mode="w", encoding="utf-8") as f:
    json.dump(seznam_studentu, f, indent=4, ensure_ascii=False)

with open("seznam.json", mode="r", encoding="utf-8") as f:
    data = json.load(f)

for student in data:
    print(f"Student {student['jmeno']} ma prumer {student['avg_znamky']}.")
```

**2. PICKLE (Serializace a deserializace)**

```python
import pickle

class Student:
    def __init__(self, jmeno, prijmeni):
        self.jmeno = jmeno
        self.prijmeni = prijmeni

    def __str__(self):
        return f"Student {self.jmeno} {self.prijmeni} je tu"

pepik = Student("Pepa", "Proutnik")
print(pepik)

pickle.dump(pepik, open('student.txt', 'wb'))

pepik_loaded = pickle.load(open('student.txt', 'rb'))
print(pepik_loaded)
```
