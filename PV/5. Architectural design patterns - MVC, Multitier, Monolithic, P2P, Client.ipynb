# 5. Architectural design patterns - MVC, Multitier, Monolithic, P2P, Client/Server

### MVC (Model-View-Controller)
* Vzor využívaný při tvorbě uživatelských rozhraní a webů
* Odděluje data, logiku a prezentaci
* Model - Stará se o data a komunikaci s databází
* View - Stará se o prezentaci dat uživateli
* Controller - Obsahuje aplikační logiku, přijímá příkazy od uživatele a propojuje Model s View

### Monolithic architecture (Monolit)
* Celá aplikace sestavena jako jeden velký a nedělitelný celek
* Lze nasadit bez sítě mezi moduly
* Kednotlivé části programu komunikují napřímo v operační paměti

### Multitier / ThreeTier
* Rozděluje aplikaci do oddělených fyzických a logických vrstev
* 3-tier architektura odděluje prezentační vrstvu (UI), aplikační vrstvu (business logiku) a datovou vrstvu

### P2P (Peer-to-Peer)
* Decentralizovaná síťová architektura
* Funguje bez centrálního serveru
* Jednotlivé uzly jsou si rovny a komunikují napřímo

### Client/Server
* Centralizovaná síťová architektura
* Klienti zasílají požadavky (requests) a centrální Server je vyřizuje a vrací odpovědi (responses)
* Pokud hlavní server spadne celá síť přestane fungovat

**UKÁZKA ARCHITEKTONICKÉHO VZORU MVC**

**1. MODEL**

```python
class UzivatelModel:
    def __init__(self, jmeno, vek):
        self.jmeno = jmeno
        self.vek = vek
```

**2. VIEW**

```python
class UzivatelView:
    def zobraz_detail_uzivatele(self, jmeno, vek):
        print("=== PROFIL UŽIVATELE ===")
        print(f"Jméno: {jmeno}")
        print(f"Věk: {vek} let")
        print("========================")
```

**3. CONTROLLER**

```python
class UzivatelController:
    def __init__(self, model, view):
        self.model = model
        self.view = view

    def aktualizuj_zobrazeni(self):
        # Controller si vezme data z Modelu a nařídí View jejich vykreslení
        self.view.zobraz_detail_uzivatele(self.model.jmeno, self.model.vek)
```

```python
# Inicializace dat
model = UzivatelModel("Karel Novák", 25)
# Inicializace vizuálu
view = UzivatelView()
# Propojení logikou
controller = UzivatelController(model, view)

# Uživatelský podnět aplikaci, aby vykreslila výsledek
controller.aktualizuj_zobrazeni()
```
