# 5. Architectural design patterns - MVC, Multitier, Monolithic, P2P, Client/Server

### MVC (Model-View-Controller)
* Vzor využívaný při tvorbě uživatelských rozhraní a webů
* Odděluje data, logiku a prezentaci
* Model - Stará se o data a komunikaci s databází
* View - Stará se o prezentaci dat uživateli
* Controller - Obsahuje aplikační logiku, přijímá příkazy od uživatele a propojuje Model s View

### Monolithic architecture (Monolit)
* Celá aplikace sestavena jako jeden velký a nedělitelný celek
* Lze nasadit bez sítě mezi moduly
* Jednotlivé části programu komunikují napřímo v operační paměti

### Multitier / ThreeTier
* Rozděluje aplikaci do oddělených fyzických a logických vrstev
* 3-tier architektura odděluje prezentační vrstvu (UI), aplikační vrstvu (business logiku) a datovou vrstvu

### P2P (Peer-to-Peer)
* Decentralizovaná síťová architektura
* Funguje bez centrálního serveru
* Jednotlivé uzly jsou si rovny a komunikují napřímo

### Client/Server
* Centralizovaná síťová architektura
* Klienti zasílají požadavky (requests) a centrální Server je vyřizuje a vrací odpovědi (responses)
* Pokud hlavní server spadne celá síť přestane fungovat

**Monolith**

```python
class Aplikace:

    def __init__(self):
        self.list_zbozi = {
            1: "nanuk",
            2: "cola",
            3: "dort"
        }
        self.objednavky = []

    def run(self):
        while True:
            print("Co chcete koupit? Vyberte z listu") #UI
            for zbozi in self.list_zbozi:
                print(zbozi, self.list_zbozi[zbozi])
            vyber = input()

            if vyber == "nic":
                print("Konec")
                break

            print(f"Vybrali jste {vyber}")
            if vyber not in self.list_zbozi.values(): #Logika
                print("Nemamem takovy zbozi, vyberte z listu")
                continue
            self.zpracuj_objednavku(vyber)

    def zpracuj_objednavku(self, zbozi):
        self.objednavky.append(zbozi) #Data
        print(f"Objednavka {zbozi} ulozena")

if __name__ == "__main__":
    app = Aplikace()
    app.run()
```

**Three-tier**

```python
class DatovaVrstva:
    def __init__(self):
        self.list_zbozi = {
            1: "nanuk",
            2: "cola",
            3: "dort"
        }
        self.objednavky = []

    def katalog(self):
        return self.list_zbozi

    def uloz(self, zbozi):
        self.objednavky.append(zbozi)
        return True

class LogickaVrstva:
    def __init__(self):
        self.db = DatovaVrstva()

    def ziskej_nabidku(self):
        return self.db.katalog()

    def over_a_zpracuj(self, vyber):
        katalog = self.db.katalog()

        if vyber not in katalog.values():
            return "Nemáme takové zboží, vyberte z listu."

        if self.db.uloz(vyber):
            return f"Objednávka {vyber} uložena."

        return "Nekde nastala chyba"

    def konec(self, vyber):
        return vyber == "nic"

class PrezentacniVrstva:
    def __init__(self):
        self.logika = LogickaVrstva()

    def run(self):
        while True:
            print("\nCo chcete koupit? Vyberte z listu:")

            nabidka = self.logika.ziskej_nabidku()
            for id_zbozi, nazev in nabidka.items():
                print(f"{id_zbozi}: {nazev}")

            vyber = input()

            if self.logika.konec(vyber):
                print("Konec")
                break

            print(f"Vybrali jste: {vyber}")

            zprava = self.logika.over_a_zpracuj(vyber)

            print(zprava)


if __name__ == "__main__":
    app = PrezentacniVrstva()
    app.run()
```
