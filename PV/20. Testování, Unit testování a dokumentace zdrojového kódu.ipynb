# 20. Testování, Unit testování a dokumentace zdrojového kódu

### Proces testování a Test Cases
* Slouží k ověřování, zda software funguje podle zadání
* Pro strukturovaný přístup se tvoří tzv. Test Cases
* Základní struktura testovacího případu obsahuje: Test Case ID, Description (popis), Pre-requisites, Test data a Expected results
* Základní metody testování jsou Black-Box (tester nevidí kód a neví jak to běží, testuje funkcionalitu) a White-Box (tester může vidět kód a ví jak funguje, testuje to jak funguje kód)  testing

### Defekty a Bug Reporty
* Pokud se skutečný výsledek neshoduje s očekávaným, vzniká softwarová chyba
* Incident chyby se sepisuje do Bug Reportu, plus přesné kroky k jejímu opětovnému vyvolání
* Ke každé chybě se definuje Severity a Priority

### Unit testing
* Automatizované ověření nejjednodušší testovatelné funkcionality izolovaně od zbytku aplikace
* V Pythonu vestavěný modul `unittest`
* K ověřování správnosti výsledků se používají asertivní metody
* Např. `self.assertEqual(a, b)`, `self.assertRaises(Vyjimka)`, ...

### Dokumentace zdrojového kódu
* Komentáře - vkládají se pomocí znaku (v Pythonu `#`), slouží především pro vývojáře ke stručnému vysvětlení konkrétního řádku kódu
* Docstringy - Formální dokumentace kódu zabalená do uvozovek (v Pythonu `"""Dokumentace ..."""`), vkládá se pod definici funkce nebo třídy, popisuje účel metody, její vstupní parametry a návratové hodnoty, moderní IDE z nich automaticky generují nápovědy, když někdo danou metodu volá
* Anotace - specifikace očekávaných datových typů (např. `def secti(a: int, b: int) -> int:`), s docstringem minimalizují nutnost studovat tělo funkce

**1. ZDROJOVÝ KÓD A JEHO DOKUMENTACE**

```python
def vydel_cisla(delenec: float, delitel: float) -> float:
    """
    Vydeli cisla mezi sebou, pokud by se snazila delit nulou tak hodi vyjimku
    :param delenec: Cislo, ktere se bude delit
    :param delitel: Cislo, kterym budeme delit
    :return: Vysledek deleni danych cisel
    """
    if delitel == 0:
        # Řádkový komentář: Ošetření chyby při dělení nulou
        raise ValueError("Nelze dělit nulou!")

    return delenec / delitel
```

**2.UNIT TESTY**

```python
import unittest
class TestMatematickychOperaci(unittest.TestCase):

    def test_spravne_deleni(self):
        self.assertEqual(vydel_cisla(10.0, 2.0), 5.0)

    def test_deleni_nulou(self):
        with self.assertRaises(ValueError):
            vydel_cisla(10, 0)

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
