# 19. Strojové učení s využitím umělých neuronových sítí

### Architektura sítě a vrstvy
* Neuronová síť (Artificial Neural Network) se skládá z propojených uzlů (neuronů)
* Ty se organizují do tří typů vrstev:
    * Vstupní vrstva (Input layer) - surová data, počet neuronů odpovídá počtu vstupních proměnných
    * Skryté vrstvy (Hidden layers) - výpočty a hledají v datech složité nelineární vzory, může jich být libovolné množství
    * Výstupní vrstva (Output layer) - konečný výsledek

##### Matematika neuronu - každé spojení mezi neurony má svou váhu (weight) určující důležitost, uvnitř neuronu se vstupy vynásobí vahami a sečtou, k výsledku se přičte prahová hodnota Bias

##### Aktivační funkce - určuje, podobu v jaké předá neuron vypočtený signál dál, vnáší do sítě nelinearitu
* ReLU - propustí pouze kladné hodnoty, záporné změní na nulu, nejčastější ve skrytých vrstvách
* Sigmoid - převádí hodnoty do rozsahu 0 až 1, ve výstupní vrstvě pro binární klasifikaci (určuje pravděpodobnost)

### Trénování
* Ztrátová funkce (Loss function) - měří odchylku mezi predikcí sítě a správným výsledkem v trénovacích datech
* Zpětné šíření chyby (Backpropagation) - algoritmus, po výpočtu chyby projde síť pozpátku a zjistí, jak moc se každý neuron podílel na celkové chybě
* Optimalizátor (např. Adam nebo Gradient Descent) - na základě backpropagation jemně upraví váhy a biasy tak, aby v dalším kroku byla chyba menší.

##### Epochy a Batches - trénování probíhá v cyklech, jedna Epocha znamená, že síť zpracovala celou sadu trénovacích dat, data se k síti neposílají najednou, ale po blocích zvaných Batches

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

**1. PŘÍPRAVA DAT**

```python
# Vstupy (dva příznaky) a očekávané výstupy (binární 0 nebo 1)
X_train = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y_train = np.array([[0], [1], [1], [0]])
```

**2. ARCHITEKTURA NEURONOVÉ SÍTĚ**

```python
model = Sequential()
# Vstupní a první skrytá vrstva: 4 neurony, vstupní dimenze=2, aktivační funkce ReLU
model.add(Dense(4, input_dim=2, activation='relu'))
# Výstupní vrstva: 1 neuron (vrací výsledek), aktivační funkce Sigmoid (pravděpodobnost 0-1)
model.add(Dense(1, activation='sigmoid'))
```

**3. KOMPILACE MODELU**

```python
# Nastavení ztrátové funkce (binary_crossentropy pro 0/1) a optimalizátoru (adam)
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

**4. TRÉNOVÁNÍ SÍTĚ**

```python
# Síť projde trénovací data 500x (epochs)
print("Začíná trénování sítě...")
model.fit(X_train, y_train, epochs=500, verbose=0)
print("Trénování dokončeno.")
```

**5. PREDIKCE A TESTOVÁNÍ**

```python
predikce = model.predict(X_train)
print("\nSurové predikce (pravděpodobnosti blížící se 0 nebo 1):")
print(predikce)

print("\nZaokrouhlený výsledek sítě (Klasifikace):")
print(np.round(predikce))
```
