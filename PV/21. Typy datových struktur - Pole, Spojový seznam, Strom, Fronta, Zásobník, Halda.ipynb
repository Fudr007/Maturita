# 21. Typy datových struktur - Pole, Spojový seznam, Strom, Fronta, Zásobník, Halda

### Pole (Array / List)
* Ukládá prvky v souvislém bloku paměti za sebou
* Okamžitý přístup k prvku přes číselný index s nejlepší možnou časovou složitostí $O(1)$
* Nevýhodou je jakákoliv změna uvnitř, při vkládání nebo mazání prvku v poli se musí všechny následující prvky v paměti posouvat, což vede ke složitosti $O(n)$

### Spojový seznam (Linked List)
* Na rozdíl od pole netvoří souvislý blok paměti
* Skládá se z nezávislých paměťových bloků (uzlů - Nodes)
* Každý uzel nese samotná data a zároveň referenci (ukazatel) na další uzel v řadě
* Obousměrná verze má referenci i na předchozí prvek
* Nelze u něj přistupovat k prvkům přes index
* Hledání prvku vyžaduje sekvenční průchod (skákání po ukazatelích) od začátku až k cíli, složitost je $O(n)$
* Výhoda: vložení nového prvku doprostřed seznamu je okamžité $O(1)$, stačí přepojit ukazatele sousedních uzlů bez posouvání dat

### Strom (Tree)
* Nelineární hierarchická struktura, která řeší pomalé vyhledávání v seznamech
* Skládá se z kořene (Root), ten se větví do poduzlů a končí listy (Leaves)
* Běžným případem je Binární vyhledávací strom, kde každý uzel má nanejvýš dva potomky (levý uzel menší než rodič, pravý je větší)
* Pokud je strom vyvážený, půlí se při každém kroku vyhledávání velikost zbývajícího prostoru k prohledání
* Časová složitost hledání, vkládání i mazání je tak logaritmická $O(\log n)$

### Fronta (Queue)
* Princip FIFO
* První prvek, který je do struktury vložen, je z ní také jako první odebrán
* Nezbytná při paralelním programování jako komunikační kanál pro návrhový vzor Producent-Konzument
* Tvoří základ algoritmu procházení grafu/stromu do šířky (BFS)

### Zásobník (Stack)
* Princip LIFO
* Poslední vložený prvek je jako první odebrán
* Procesory a běhová prostředí využívají takzvaný Call Stack (zásobník volání) k pamatování si návratových adres při zanořování funkcí nebo u rekurze
* Základem algoritmu prohledávání do hloubky (DFS)

### Halda (Heap)
* Specializovaná stromová struktura, využívaná převážně k vytvoření takzvané Prioritní fronty
* Kořenový uzel (prvek s indexem 0) představuje absolutní minimum (v případě Min-Haldy) nebo maximum (Max-Haldy) z celé struktury
* Čtení nejmenšího prvku je okamžité $O(1)$
* Při odebrání se strom automaticky a efektivně přeskládá v čase $O(\log n)$, aby se na vrchol okamžitě posunulo další aktuální minimum/maximum

```python
import collections
import heapq
```

**1. POLE VS. ZÁSOBNÍK**

```python
# V Pythonu se zásobník implementuje elegantně přímo pomocí pole (listu)
zasobnik = []
zasobnik.append("První vloženo")
zasobnik.append("Druhé vloženo")
zasobnik.append("Poslední vloženo")

# Metoda pop() vezme a odstraní prvek z konce (O(1) operace)
print("Vyjmuto ze zásobníku (LIFO):", zasobnik.pop()) # Vypíše: Poslední vloženo
```


**2. FRONTA**

```python
# Použití pole by bylo pro frontu neefektivní, proto se používá oboustranná fronta (deque)
fronta = collections.deque()
fronta.append("Zákazník 1")
fronta.append("Zákazník 2")
fronta.append("Zákazník 3")

# Metoda popleft() vezme a odstraní prvek ze začátku (O(1) operace)
print("Vyjmuto z fronty (FIFO):", fronta.popleft()) # Vypíše: Zákazník 1
```

**3. SPOJOVÝ SEZNAM (LINKED LIST)**

```python
class Uzel:
    def __init__(self, data):
        self.data = data
        self.dalsi = None # Ukazatel (reference) na další uzel

# Vytvoření nezávislých uzlů v paměti
prvni = Uzel("A")
druhy = Uzel("B")
treti = Uzel("C")

# Ruční přepojení referencí tvořící spojový seznam
prvni.dalsi = druhy
druhy.dalsi = treti

print("Hodnota dalšího uzlu od prvního:", prvni.dalsi.data)
```

**4. HALDA**

```python
# V Pythonu se halda ukládá pod kapotou do normálního pole,
# ale operace nad ním řídí speciální algoritmy modulu heapq
halda = []
# Vkládání prvků v náhodném pořadí (Složitost O(log n))
heapq.heappush(halda, 50)
heapq.heappush(halda, 10)
heapq.heappush(halda, 30)

print("Podkladové pole haldy po přidání:", halda) # [10, 50, 30] - Na indexu 0 je vždy minimum
# Bez ohledu na to, jak prvky vkládáme, pop() vyjme ten absolutně nejmenší
print("Nejmenší prvek (Min-Heap):", heapq.heappop(halda))
```

**5. STROM**

```python
class UzelStromu:
    def __init__(self, klic):
        self.klic = klic
        # Dvě specifické reference místo jedné
        self.levy_potomek = None  # Cesta k menším prvkům
        self.pravy_potomek = None # Cesta k větším prvkům

koren = UzelStromu(50)
koren.levy_potomek = UzelStromu(20)
koren.pravy_potomek = UzelStromu(70)
print(f"Kořen: {koren.klic}, menší potomek: {koren.levy_potomek.klic}, větší: {koren.pravy_potomek.klic}")
```
