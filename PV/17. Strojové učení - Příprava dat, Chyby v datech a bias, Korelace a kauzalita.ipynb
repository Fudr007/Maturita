# 17. Strojové učení - Příprava dat, Chyby v datech a bias, Korelace a kauzalita

### Příprava dat
* Data samotná jsou jen zaznamenané údaje
* Informací se data stávají až dodáním kontextu a významu
* Pro strojové učení je dělíme na:
    * Strukturovaná: Počítačem snadno zpracovatelná, jasná struktura (např. relační databáze nebo CSV soubory)
    * Nestrukturovaná: Pro člověka srozumitelná, pro počítač složitější (např. obrázky, videa, volný text)
* Data se získávají měřením (senzory), průzkumy, ruční tvorbou nebo generováním

### Reprezentativita a šum
* Pokud máme dat málo nebo neodpovídají zkoumanému problému, nejsou reprezentativní a nelze z nich vyvozovat závěry
* Šum je náhodná chyba vyskytující se v datech, pokud není příliš výrazný tak většině algoritmů nevadí

### Bias
* Narozdíl od šumu je bias systematická chyba
* Ta je velmi nebezpečná, protože přímo a výrazně ovlivňuje globální statistiky
* Způsobuje ji např. výběrové zkreslení, typicky třeba počítání průměrné mzdy v ČR z datového vzorku, ve kterém jsou jen absolventi vysokých škol žijících v Praze

### Korelace a kauzalita
* Problémem při interpretaci dat je záměna těchto dvou pojmů
* Korelace vyjadřuje statistickou závislost mezi dvěma veličinami, např. podle dat platí, že vysoký výskyt žížal na zahradě silně koreluje s vysokým výskytem slimáků
* Kauzalita je příčinná souvislost, kde jedna událost přímo způsobuje druhou, např. výskyt žížal nezpůsobuje výskyt slimáků a naopak, společnou příčinou je deštivé počasí
* Model strojového učení najde korelaci, kauzalitu musí pohlídat datový vědec

```python
import pandas as pd
```

**1. PŘÍPRAVA DAT (STRUKTUROVANÁ DATA)**

```python
# Sestavení tabulkových dat (features) pro trénování nebo analýzu
data = {
    'Mzda': [30000, 35000, 40000, 45000, 150000, 32000, 38000],
    'Vek': [25, 28, 32, 35, 55, 26, 30],
    'Roky_praxe': [2, 5, 8, 10, 30, 3, 6]
}
df = pd.DataFrame(data)
```

**2. CHYBY V DATECH: VÝBĚROVÝ BIAS**

```python
skutecny_prumer = df['Mzda'].mean()
print(f"Skutečná průměrná mzda populace: {skutecny_prumer:.2f} Kč")

# Simulace výběrového zkreslení (Bias):
# Datový analytik pro model omylem vyfiltroval pouze vzorek lidí s praxí nad 10 let
bias_vzorek = df[df['Roky_praxe'] >= 10]
bias_prumer = bias_vzorek['Mzda'].mean()
print(f"Zkreslená mzda (Bias vzorek - jen experti): {bias_prumer:.2f} Kč")
print("-> Výsledek modelu bude kvůli výběrové chybě silně nadhodnocený.\n")
```

**3. KORELACE**

```python
# Automatický výpočet korelační matice mezi všemi proměnnými (-1.0 až 1.0)
korelace = df.corr()
print("Korelační matice:")
print(korelace)

# Zjištění míry korelace pouze mezi dvěma specifickými jevy
kor_mzda_praxe = df['Mzda'].corr(df['Roky_praxe'])
print(f"\nKorelace mezi mzdou a roky praxe: {kor_mzda_praxe:.3f}")
# Vysoká hodnota blížící se 1 ukazuje silnou korelaci.
# Zda jde o kauzalitu, musí analytik potvrdit vlastním zkoumáním (zde to smysl dává).
```
