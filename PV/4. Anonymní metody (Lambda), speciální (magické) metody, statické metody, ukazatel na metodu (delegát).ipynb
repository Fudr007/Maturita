# 4. Anonymní metody (Lambda), speciální (magické) metody, statické metody, ukazatel na metodu (delegát)

### Anonymní metody (Lambda funkce)
* Zkrácený zápis jednoduché funkce
* Nemá přidělené žádné jméno
* Definují se klíčovým slovem `lambda`
* Mohou přijímat libovolný počet parametrů
* Smí obsahovat pouze jeden výraz, jehož výsledek automaticky vrací
* Tam kde potřebujeme jednorázovou funkci na jeden řádek
* Pro řazení složitějších struktur nebo pro funkce `map()` a `filter()`

### Speciální/Dunder metody
* Umožňují objektům interagovat s vestavěnými operátory a funkcemi daného jazyka
* V Pythonu se označují jako dunder
* Běžné příklady:
    * `__init__()` - Konstruktor objektu, volá se při jeho vytvoření
    * `__str__()` - Jak se má objekt vypsat, když na něj zavoláme `print()`
    * `__eq__()` a `__hash__()` - Slouží k porovnání dvou objektů přes operátor == a zjištění jejich unikátnosti
    * `__add__()` - Slouží k přetěžování operátorů, určuje, co se stane, když mezi dva objekty napíšeme znaménko `+`, s jiným názvem se dělají i jiná znaménka

### Statické metody
* Nepatří instanci, ale samotné třídě
* V Pythonu se označují `@staticmethod` a nepřijímají referenci na objekt
* Pro pomocné výpočty logicky patřící k dané třídě, ale nepotřebují znát ani měnit stav konkrétního objektu

### Ukazatel na metodu (Delegát)
* V kompilovaných jazycích slouží jako typově bezpečné ukazatele na funkce, umožňuje to předávat metody jako parametry jiným metodám
* V dynamických jazycích žádný speciální datový typ "delegát" neexistuje, funkce jsou zde objekty první třídy
* Samotný název funkce (bez závorek) funguje jako přímý ukazatel do paměti, můžeme ho uložit do proměnné, poslat jako argument jiné funkci nebo ho vrátit

**1. SPECIÁLNÍ A STATICKÉ METODY**

```python
class Vektor:
    def __init__(self, x, y):
        # Magická metoda: Konstruktor
        self.x = x
        self.y = y

    def __str__(self):
        # Magická metoda: Převede objekt na hezky čitelný text
        return f"Vektor[{self.x}, {self.y}]"

    def __add__(self, jiny_vektor):
        # Magická metoda: Přetížení operátoru +. Umožní sečíst dva objekty Vektor
        nove_x = self.x + jiny_vektor.x
        nove_y = self.y + jiny_vektor.y
        return Vektor(nove_x, nove_y)

    @staticmethod
    def popis_tridy():
        # Statická metoda: Nemá parametr 'self', volá se přímo přes třídu
        return "Třída Vektor slouží k matematickým operacím ve 2D prostoru."


# Testování magických a statických metod
v1 = Vektor(1, 2)
v2 = Vektor(3, 4)
v3 = v1 + v2 # Zde se na pozadí volá magická metoda __add__

print("1. Volání magických metod (+ a print):", v3)
print("2. Volání statické metody:", Vektor.popis_tridy())
```

**2. UKAZATEL NA METODU (DELEGÁT)**

```python
def zpracuj_operaci(ukazatel_na_funkci, hodnota1, hodnota2):
    # Funkce přijme jako parametr "delegáta" (referenci na jinou funkci)
    print("3. Zpracovávám dodanou funkci přes ukazatel...")
    vysledek = ukazatel_na_funkci(hodnota1, hodnota2)
    return vysledek
```

**3. ANONYMNÍ METODY (LAMBDA)**

```python
# Vytvoření anonymní funkce a její okamžité uložení do proměnné
nasobeni = lambda a, b: a * b

# Předání lambda funkce i standardní funkce přes UKAZATEL do jiné metody
vysledek_lambdy = zpracuj_operaci(nasobeni, 10, 5)
print("Výsledek lambdy:", vysledek_lambdy)

# Lambdy se často předávají přímo (bez uložení) jako argumenty (např. klíč pro řazení)
zbozi = [("Jablka", 50), ("Maso", 200), ("Banány", 30)]
# Chceme seřadit seznam N-tic podle ceny (index 1). Předáme lambdu jako ukazatel 'key'.
zbozi.sort(key=lambda polozka: polozka[1])
print("4. Seřazeno pomocí anonymní lambda funkce:", zbozi)
```
