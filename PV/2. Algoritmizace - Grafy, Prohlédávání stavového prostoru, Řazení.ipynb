# 2. Algoritmizace - Grafy, Prohlédávání stavového prostoru, Řazení

### Grafy
* Abstraktní datová struktura, slouží k reprezentaci sítě vztahů
* Skládá se z uzlů (vrcholů), ty představují entity, a hran definujících vztahy mezi nimi
* Typy hran:
    * orientované - jednosměrka
    * neorientované - obousměrka
    * ohodnocené - jaká je vzdálenost, nebo kolik stojí jí přejít
    * neohodnocené
* Reprezentace v paměti: Nejčastěji se graf ukládá jako Matice sousednosti nebo jako Seznam sousedů

### Prohledávání stavového prostoru
* Způsob, jak v grafu nebo stromu najít cílový uzel nebo nejkratší cestu
* BFS (Breadth-First Search) - Postupuje po vrstvách, algoritmus využívá datovou strukturu Fronta, v neohodnoceném grafu garantuje nalezení nejkratší cesty
* DFS (Depth-First Search) - Vydá se jednou cestou do maximální možné hloubky, teprve když narazí na slepou uličku, začne se vracet zpět a zkusí jinou větev, využívá Zásobník nebo staví na Rekurzi, paměťově je méně náročný než BFS, ale negarantuje nejkratší cestu

### Řazení
* Proces uspořádání prvků v poli
* Jednoduché algoritmy:
    * Pro velká data nepoužitelné, ale snadno se programují
    * Bubble Sort - Opakovaně porovnává sousední prvky a prohazuje je, pokud jsou špatně
    * Insertion Sort - Bere nevyřešené prvky a vkládá je na správné místo do již seřazené levé části pole
* Pokročilé algoritmy:
    * Využívají přístup "Rozděl a panuj"
    * Quick Sort - Zvolí si Pivota, pole rozdělí na prvky menší než pivot a větší než pivot, následně se rekurzivně opakuje pro obě poloviny, v praxi jeden z nejrychlejších algoritmů
    * Merge Sort - Rozpůlí pole až na jednotlivé prvky a následně je postupně slučuje zpět do správného pořadí

**1. GRAFY A PROHLEDÁVÁNÍ STAVOVÉHO PROSTORU (BFS)**

```python
from collections import deque

graf = {
    'Start': ['A', 'B'],
    'A': ['Start', 'C', 'D'],
    'B': ['Start', 'E'],
    'C': ['A'],
    'D': ['A', 'Cil'],
    'E': ['B'],
    'Cil': ['D']
}

def najdi_cestu_bfs(graf, start, cil):
    fronta = deque([start])

    predchudce = {start: None}

    print(f"Hledani z {start} do {cil}")

    while fronta:
        uzel = fronta.popleft()

        if uzel == cil:
            cesta = []
            krok = cil

            while krok is not None: #od konce jdeme pokrocich, kterymi jsme prisli
                cesta.append(krok)
                krok = predchudce[krok]

            cesta.reverse()
            return f"Cil nalezen, cesta: {cesta}"

        for soused in graf[uzel]:
            if soused not in predchudce:
                predchudce[soused] = uzel
                fronta.append(soused)

    return "Cesta nenalezena"

najdi_cestu_bfs(graf, 'Start', 'Cil')
```

2. ŘAZENÍ (QUICK SORT)

```python
def quick_sort(pole):
    if len(pole) <= 1:
        return pole

    pivot = pole[len(pole) // 2]
    print(f"Pivot: {pivot}")

    mensi = [x for x in pole if x < pivot]
    rovno = [x for x in pole if x == pivot]
    vetsi = [x for x in pole if x > pivot]

    return quick_sort(mensi) + rovno + quick_sort(vetsi) #rekuze a deleni na mensi casti dokud neni pole jedno-prvkove

data = [10,6,7,50,12,32,37,52,10,1]
print(f"Neserazena {data}")
print(f"Serazena {quick_sort(data)}")
```
