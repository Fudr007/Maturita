# 3. Algoritmizace - Rekurze, Brute Force, Heuristiky, Nedeterministické algoritmy

u### Rekurze
* Funkce v rámci svého těla volá sama sebe
* Musí zde být ukončovací podmínka (Base case), jinak se Stack zaplní a program spadne na chybu StackOverflow
* Využívá se pro procházení stromových struktur nebo generování všech možných kombinací a permutací

### Brute Force
* Systematicky prohledá kompletně celý stavový prostor
* Najde všechna řešení a tudíž i absolutně nejlepší řešení
* Často exponenciální $O(2^n)$ nebo faktoriálová časová složitost $O(n!)$

### Heuristiky
* Snaží se najít dostatečně řešení ve velmi krátkém čase díky nějakému triku
* Zmenšuje počet nutných kombinací k prohledání stavového prostoru

### Nedeterministické algoritmy
* Například Monte Carlo algoritmus
* Úmyslně zaveden prvek náhody
* Pro stejná vstupní data mohou při každém spuštění vrátit jiný výsledek nebo běžet jinou dobu
* Negarantuje nejoptimálnější řešení
* Provede sérii náhodných pokusů a z nich vybere ten nejlepší výsledek

**Rekurze**

```python
def faktorial(n, hloubka = 0):

    odsazeni = "-" * hloubka
    print(f'{odsazeni} volani faktorialu {n}')

    if n<=1:
        return 1 #ukoncovaci podminka

    vysledek = n * faktorial(n-1, hloubka+1)

    print(f'{odsazeni} vysledek faktorialu {n} * {n-1}! je {vysledek}')
    return vysledek

print(faktorial(5))
```

**Brute Force**

```python
def prolomeni_force(pin):
    pocet_pokusu = 0
    for i in range(10000):
        pocet_pokusu += 1

        tip = str(i).zfill(4)

        if tip == pin:
            return pocet_pokusu
```

**Heuristika**

```python
def prolomeni_heuristika(pin):
    pocet_pokusu = 0

    logicke_piny = ["1234", "4321", "0000", "1111", "2222", "3333", "4444", "5555", "6666", "7777", "8888", "9999"] + [str(rok) for rok in range(1900, 2025)]

    for tip_pinu in logicke_piny:
        pocet_pokusu += 1
        if tip_pinu == pin:
            return pocet_pokusu

    return "Divna kombinace"
```

**Monte Carlo**

```python
import random

def prolomeni_random(pin):
    pocet_pokusu = 0

    while True:
        pocet_pokusu += 1
        tip = str(random.randint(0, 9999)).zfill(4)

        if tip == pin:
            return pocet_pokusu
```

**Srovnani**

```python
print(f"Hledany pin je {tajny_pin}")
print(f"Prolomeni force: {prolomeni_force(tajny_pin)}")
print(f"Prolomeni heuristika: {prolomeni_heuristika(tajny_pin)}")
print(f"Prolomeni random: {prolomeni_random(tajny_pin)}")
```
