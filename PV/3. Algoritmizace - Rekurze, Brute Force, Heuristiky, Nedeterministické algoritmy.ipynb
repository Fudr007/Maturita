# Algoritmizace - Rekurze, Brute Force, Heuristiky, Nedeterministické algoritmy

### Rekurze
* Funkce v rámci svého těla volá sama sebe
* Musí zde být ukončovací podmínka (Base case), jinak se Stack zaplní a program spadne na chybu StackOverflow
* Využívá se pro procházení stromových struktur nebo generování všech možných kombinací a permutací

### Brute Force
* Systematicky prohledá kompletně celý stavový prostor
* Najde všechna řešení a tudíž i absolutně nejlepší řešení
* Často exponenciální $O(2^n)$ nebo faktoriálová časová složitost $O(n!)$

### Heuristiky
* Snaží se najít dostatečně řešení ve velmi krátkém čase díky nějakému triku
* Zmenšuje počet nutných kombinací k prohledání stavového prostoru

### Nedeterministické algoritmy
* Například Monte Carlo algoritmus
* Úmyslně zaveden prvek náhody
* Pro stejná vstupní data mohou při každém spuštění vrátit jiný výsledek nebo běžet jinou dobu
* Negarantuje nejoptimálnější řešení
* Provede sérii náhodných pokusů a z nich vybere ten nejlepší výsledek

```python
import itertools
import random
import math
```

```python
# Vzdálenosti mezi 4 body (např. A, B, C, D)
vzdalenosti = {
    ('A', 'B'): 10, ('A', 'C'): 15, ('A', 'D'): 20,
    ('B', 'C'): 35, ('B', 'D'): 25, ('C', 'D'): 30,
    # Zpětné cesty pro zjednodušení simulujeme stejně
    ('B', 'A'): 10, ('C', 'A'): 15, ('D', 'A'): 20,
    ('C', 'B'): 35, ('D', 'B'): 25, ('D', 'C'): 30
}

mesta = ['A', 'B', 'C', 'D']

def spocitej_delku_trasy(trasa):
    delka = 0
    for i in range(len(trasa) - 1):
        delka += vzdalenosti[(trasa[i], trasa[i+1])]
    return delka
```

**1. REKURZE A BRUTE FORCE**

```python
# itertools.permutations pod kapotou využívá rekurzi ke generování VŠECH tras O(n!)
nejlepsi_trasa_bf = None
nejkratsi_vzdalenost_bf = float('inf')

for trasa in itertools.permutations(mesta):
    delka = spocitej_delku_trasy(trasa)
    if delka < nejkratsi_vzdalenost_bf:
        nejkratsi_vzdalenost_bf = delka
        nejlepsi_trasa_bf = trasa

print(f"Brute Force (100% optimální): {nejlepsi_trasa_bf} | Délka: {nejkratsi_vzdalenost_bf}")
```

**2. HEURISTIKA**

```python
# Začneme v náhodném městě a vždy jdeme do toho aktuálně nejbližšího
start_mesto = 'A'
nenavstivena = ['B', 'C', 'D']
heuristicka_trasa = [start_mesto]

aktualni = start_mesto
while nenavstivena:
    # Vybere souseda s nejmenší vzdáleností (lokální optimum)
    dalsi_mesto = min(nenavstivena, key=lambda m: vzdalenosti[(aktualni, m)])
    heuristicka_trasa.append(dalsi_mesto)
    nenavstivena.remove(dalsi_mesto)
    aktualni = dalsi_mesto

print(f"Heuristika (Velmi rychlé): {heuristicka_trasa} | Délka: {spocitej_delku_trasy(heuristicka_trasa)}")
```

**3. NEDETERMINISTICKÝ PŘÍSTUP**

```python
# Uděláme N náhodných pokusů a vybereme ten nejlepší.
nejlepsi_trasa_mc = None
nejkratsi_vzdalenost_mc = float('inf')

pocet_pokusu = 5
for _ in range(pocet_pokusu):
    # Náhodné zamíchání vnáší nedeterminismus
    nahodna_trasa = mesta.copy()
    random.shuffle(nahodna_trasa)

    delka = spocitej_delku_trasy(nahodna_trasa)
    if delka < nejkratsi_vzdalenost_mc:
        nejkratsi_vzdalenost_mc = delka
        nejlepsi_trasa_mc = nahodna_trasa

print(f"Monte Carlo ({pocet_pokusu} pokusů): {nejlepsi_trasa_mc} | Délka: {nejkratsi_vzdalenost_mc}")
```
