# 15. Programovací jazyky - vlastnosti, srovnání, popis způsobu tvorby i běhu programů

### Klasifikace programovacích jazyků
* Low-level - Jazyky blízké hardwaru, Strojový kód, Assembler, závislé na architektuře procesoru, velmi rychlé, ale náročné na psaní
* High-level - Jazyky s vysokou mírou abstrakce, Python, Java, C#, lépe čitelné pro člověka a přenositelné mezi různými systémy

### Způsoby tvorby a běhu programů
* Kompilace - Zdrojový kód jednorázově přeložen kompilátorem do strojového kódu pro konkrétní procesor, odhalí syntaktické chyby před spuštěním, výsledný program je velmi rychlý, např. C, C++, Rust
* Interpretace - Zdrojový kód čte a vykonává interpret řádek po řádku přímo za běhu, snadné ladění a nezávislost na platformě, např. JavaScript, PHP
* Hybridní přístup - Kód je nejprve přeložen do mezikódu (Bytecode), poté vykonáván virtuálním strojem, často se využívá JIT kompilace (kritické části kódu přeloží až v okamžiku potřeby)

### Typování
* Statické vs. Dynamické - Statického typování (C++, Java) se typ proměnné určuje při psaní/kompilaci a nelze jej měnit, u dynamického (Python) se typ určí až podle přiřazené hodnoty za běhu
* Silné vs. Slabé - Silně typované jazyky (Python, Java) nedovolí bez explicitního převodu operace mezi nekompatibilními typy, slabě typované (C, JS) udělají automatický převod, může vést k chybám

### Správa paměti
* Manuální - Vývojář musí sám alokovat (new, malloc) a uvolňovat (delete, free) paměť
* Automatická (Garbage Collector - GC) - proces běžící na pozadí, automaticky vyhledává objekty, na které již neexistuje reference, uvolňuje jejich místo v paměti

### Srovnání vybraných jazyků
* **C++** - Kompilovaný, statický, manuální správa paměti, extrémně výkonný, vhodný pro hry a systémové nástroje
* **Java** - Hybridní, statický, automatická správa paměti, bezpečný, robustní, vhodný pro korporátní systémy a Android
* **Python** - Interpretovaný/Hybridní, dynamický, automatická správa paměti, přehledný, obrovské množství knihoven, ideální pro AI a skriptování

**2. SPRÁVA REFERENCÍ A PAMĚTI**

```python
class MujObjekt:
    def __init__(self, jmeno):
        self.jmeno = jmeno
    def __del__(self):
        print(f"Objekt '{self.jmeno}' byl odstraněn z paměti.")

# Vytvoření objektu a sledování počtu referencí
obj = MujObjekt("Maturita")
# sys.getrefcount vrací počet odkazů na objekt (včetně dočasného v této funkci)
print(f"Počet referencí na objekt: {sys.getrefcount(obj)}")

# Vytvoření další reference
druhy_odkaz = obj
print(f"Počet referencí po přidání odkazu: {sys.getrefcount(obj)}")

# Odstranění referencí
del obj
print("Smazána první reference.")
del druhy_odkaz
# Jakmile počet referencí klesne na nulu, Garbage Collector (GC) objekt uvolní.
# V Pythonu to proběhne okamžitě nebo při nejbližším průběhu GC.
gc.collect()
```
