# 8. Dědičnost, method overriding, function overloading

### Dědičnost
* Jeden z pilířů objektově orientovaného programování
* Umožňuje vytvářet nové třídy na základě již existujících tříd
* Potomek automaticky dědí všechny vlastnosti a metody svého rodiče
* Eliminuje duplicity v kódu
* Podporuje znovupoužitelnost kódu
* V Pythonu se definuje vložením rodičovské třídy do kulatých závorek

### Method overriding
* Souvisí s dědičností
* Nastává když potomek zdědí metodu, ale potřebuje, aby se chovala jinak
* V praxi se metoda v potomkovi napíše se stejným názvem s odlišným kódem
* Při volání této metody na objektu potomka se následně spustí jeho verze

### Function overloading
* V rámci jedné třídy nebo programu existuje více metod se stejným názvem, ale s odlišnou signaturou (vstupními parametry)
* Jazyky jako C#, Java nativně podporují, v Pythonu se toto dělá přes defaultní paramtetry nebo `*args`

**1. DĚDIČNOST, METHOD OVERRIDING**

```python
class Zamestnanec:
    def __init__(self, jmeno, zakladni_plat):
        self.jmeno = jmeno
        self.zakladni_plat = zakladni_plat

    # Metoda rodiče
    def vypocitej_plat(self):
        return self.zakladni_plat

# Třída Manazer dědí všechny vlastnosti od třídy Zamestnanec
class Manazer(Zamestnanec):
    def __init__(self, jmeno, zakladni_plat, bonus):
        # Volání konstruktoru rodiče
        super().__init__(jmeno, zakladni_plat)
        self.bonus = bonus

    # Potomek mění chování zděděné metody pro své vlastní potřeby
    def vypocitej_plat(self):
        # Oproti rodiči připočítává k základu i bonus
        return self.zakladni_plat + self.bonus
```

**2. FUNCTION OVERLOADING**

```python
class KalkulackaOdmen:
    # V Pythonu simulujeme odlišnou signaturu výchozí hodnotou (None)
    def pridej_odmenu(self, zaklad, extra_bonus=None):
        if extra_bonus is not None:
            # Signatura při zavolání s oběma parametry
            return zaklad + extra_bonus
        # Signatura při zavolání pouze s prvním parametrem
        return zaklad + 500
```

```python
bezny = Zamestnanec("Karel", 30000)
sef = Manazer("Jana", 40000, 15000)

print("--- Ukázka Method Overriding ---")
# Obě instance volají metodu stejného jména, ale kvůli Overridingu s jiným výsledkem
print(f"Plat běžného zaměstnance: {bezny.vypocitej_plat()} Kč")
print(f"Plat manažera: {sef.vypocitej_plat()} Kč")

print("\n--- Ukázka Function Overloading ---")
kalkulacka = KalkulackaOdmen()
# Volání se dvěma parametry
print("Odmena (2 parametry):", kalkulacka.pridej_odmenu(1000, 2000))
# Volání s jedním parametrem (Python využije výchozí chování)
print("Odmena (1 parametr):", kalkulacka.pridej_odmenu(1000))
```
