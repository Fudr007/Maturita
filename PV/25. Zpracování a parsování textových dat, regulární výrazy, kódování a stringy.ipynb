
# 25. Zpracování a parsování textových dat, regulární výrazy, kódování a stringy

### Stringy a formátování
* Pro manipulaci s textovými řetězci se používají metody formátování, např. volání metody `format()`, `strip()`
* Využívá se například pro zajištění přesného tvaru výstupu, jako je datum D. M. YYYY. Uvnitř format() je možné uplatnit specifika pro zobrazení, jako například `{:d}` pro formátování celého čísla

### Kódování (Encoding) a binární data
* V Pythonu lze přímo zapisovat literály bytů, a to pomocí předpony b před jednoduchými uvozovkami (`b'...'`)
* Do uvozovek se vkládají hodnoty v šestnáctkové soustavě po jednotlivých bytech. Např. zápis bytu s hodnotou jedna jako `b'\x01'`, nebo dvou bytů s hodnotami 0 a 255 jako `b'\x00\xFF'`
* Stejného binárního výsledku docílíme převedením celých čísel přes funkci `bytes()`, tudíž platí, že `bytes([1,2,3]) == b'\x01\x02\x03'`
* Ze stringu na byty to probíhá pomocí funkce `encode()`. Nezbytné pokud textová data potřebujeme např. kryptograficky zpracovat či hashovat (např. algoritmem SHA384)

### Parsování a zpracování textových souborů
* Čtení textového souboru je pomocí `read()` pro celý soubor, `readline()` pro řádek, iterací v cyklu `for line in file:` pro průchod všech řádků
* Data v CSV se zpracovávají iterováním přes `csv.reader`, lze specifikovat parametry jako oddělovač `delimiter=','`, znak uvozovek `quotechar='"'`, přeskočení hlavičky lze přes `next(reader, None)`.
* Pro parsování strukturovaných XML dat je např. knihovna `xml.etree.ElementTree`.

### Regulární výrazy (Regex)
* Vyhledání podle formálního vzoru textu, pro to slouží mj. kotvy (anchors) a speciální třídy znaků
* Značky: `.` nahrazuje libovolný znak (zápis `\.` je literaly ten znak), znak `$` značí konec řetězce a `^` počátek řetězce
* Regex výrazy:
    * `^x[1-5]$` - stringy začínající písmenem x a končící jedinou číslicí v rozsahu 1 až 5
    * `^(?!Species$).*` - vše kromě řetězce přesně složeného z "Species" (tzv. negativní lookahead)
* V Pandas jsou regex výrazy pro hromadné filtrování názvů sloupců (tzv. Variables), a to voláním `df.filter(regex='...')`

```python
import pandas as pd
import csv
```

**1. STRINGY A FORMÁTOVÁNÍ**

```python
den, mesic, rok = 5, 9, 2025
narozeni_str = "{:d}. {:d}. {:d}".format(den, mesic, rok)
print("Datum:", narozeni_str)
```

**2. KÓDOVÁNÍ (ENCODING) A BINÁRNÍ ZÁPIS**

```python
rodne_cislo = "990101/1234"
rodne_cislo_byty = rodne_cislo.encode()

hodnoty_literalu = b'\x00\xFF'
hodnoty_z_cisel = bytes([0, 255])
print("Je hodnota bytů stejná?", hodnoty_literalu == hodnoty_z_cisel)
```

**3. PARSOVÁNÍ TEXTOVÝCH DAT (CSV)**

```python
with open('dluznici.csv', 'r') as csvfile:
    reader = csv.reader(csvfile, delimiter=',', quotechar='"')

    next(reader, None)
    for row in reader:
        print("Záznam - ID: {}, Rodné číslo: {}, Dluh: {}".format(row[0], row[1], row[2]))
        break
```

**4. REGULÁRNÍ VÝRAZY**

```python

df = pd.DataFrame({
    "x1": [1],
    "x2": [2],
    "SepalLength": [5.1],
    "Species": ["Iris-setosa"]
})

sloupce_s_x = df.filter(regex='^x[1-5]$')
print("Filtrované sloupce x:", list(sloupce_s_x.columns))

bez_species = df.filter(regex='^(?!Species$).*')
print("Sloupce bez Species:", list(bez_species.columns))
```
