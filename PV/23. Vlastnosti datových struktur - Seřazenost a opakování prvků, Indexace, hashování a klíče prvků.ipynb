
# 23. Vlastnosti datových struktur - Seřazenost a opakování prvků, Indexace, hashování a klíče prvků


### Seřazenost
* Udává, zda si datová struktura zachovává pořadí, ve kterém do ní byly prvky vloženy
* Pole (list) a spojový seznam jsou seřazené struktury
* Pokud je seřazená struktura navíc i setříděná podle velikosti, umožňuje to použít vysoce efektivní binární hledání

### Opakování prvků
* Pole nebo spojový seznam běžně umožňují vícenásobné uložení shodného prvku
* Kolekce typu množina (set) opakování prvků neumožňuje. Jakmile do ní vložíme prvek, který už obsahuje, je prvek vyřazen jako duplicita

### Indexace
* Schopnost přímého a okamžitého přístupu k prvku prostřednictvím jeho pořadového čísla (indexu) nebo klíče
* Index umožňuje rychlejší vyhledávání prvků bez nutnosti procházet celou strukturu
* Klasický přístup přes index do pole (např. `pole[5]`) má velmi rychlou časovou složitost $O(1)$

### Hashování a klíče prvků
* Vyhledávací struktury (jako jsou slovníky/dictionary) ukládají hodnoty pod unikátními klíči
* Aby do nich byl přístup rychlý, klíč se převede takzvanou hashovací funkcí na číselný otisk (hash)
* V Pythonu je možné řídit, jaký hash objekt dostane, přetížením `__hash__()`
* Aby například kolekce typu `set()` úspěšně vyřadila vkládaný prvek jako duplicitu, musí mít nově vkládaný i již uložený objekt shodný hash, a zároveň musí operátor `==` vrátit `True`
* Může stát, že dva různé klíče dostanou matematicky shodný hash (hash kolize), výskyt následně zpomalí operace nad danou datovou strukturou

| Datová struktura v Pythonu | Zachovává pořadí vložení? | Přístup k prvkům (Indexace) | Povoluje duplicitní prvky? | Lze po vytvoření měnit? (Mutable) | Charakteristika a využití |
| :--- | :--- | :--- | :--- | :--- | :--- |
| **List** (seznam/pole)<br>`[1, 2, 2]` | Ano | Ano <br>*(číselný index)* | Ano | Ano | Univerzální struktura. Pomalé prohledávání podle hodnoty (složitost $O(n)$). |
| **Tuple** <br>`(1, 2, 2)` | Ano | Ano <br>*(číselný index)* | Ano | **Ne** <br>*(Immutable)* | Slouží pro fixní záznamy (chrání data před změnou). Lze použít jako klíč do slovníku (je hashovatelný). |
| **Dictionary** (slovník)<br>`{"a": 1, "b": 1}` | Ano <br>*(od Python 3.7+)* | Ano <br>*(přes unikátní klíč)* | **Ne** (u klíčů)<br>Ano (u hodnot) | Ano | Tabulka (hash mapa) pro mapování klíč:hodnota. Bleskové vyhledávání podle klíče ($O(1)$). |
| **Set** (množina)<br>`{1, 2, 3}` | **Ne** | **Ne** | **Ne** | Ano | Udržování pouze unikátních hodnot. Bleskové ověření, zda struktura prvek obsahuje ($O(1)$). |
| **Frozenset**<br>`frozenset([1, 2])` | **Ne** | **Ne** | **Ne** | **Ne** <br>*(Immutable)* | Neměnná verze množiny. Díky neměnnosti ji lze vložit jako prvek do jiného setu nebo použít jako klíč ve slovníku. |

**3. HASHOVÁNÍ A ZABRÁNĚNÍ OPAKOVÁNÍ PRVKŮ V SETU**

```python
class Zbozi:
    def __init__(self, nazev, vaha):
        self.nazev = nazev
        self.vaha = vaha

    # Abychom mohli objekty vkládat do setu bez opakování
    def __hash__(self):
        return hash((self.nazev, self.vaha))

    # Aby množina (set) poznala duplicitu, musí mít shodný hash musí se rovnat přes operátor == (__eq__)
    def __eq__(self, other):
        if not isinstance(other, Zbozi):
            raise Exception('Porovnávat lze jen zboží mezi sebou')
        return self.nazev == other.nazev and self.vaha == other.vaha


# Testování vložení do kolekce typu set() bez duplicit
zbozi1 = Zbozi("Mrkev", 1)
zbozi2 = Zbozi("Mrkev", 1)  # Má stejné atributy jako zbozi1 -> stejný hash
zbozi3 = Zbozi("Celer", 1)

# Vytvoření prázdného setu
unikatni_zbozi = set()

# Vložení prvků
unikatni_zbozi.add(zbozi1)
unikatni_zbozi.add(zbozi2)  # Toto bude setem ignorováno jako duplicita
unikatni_zbozi.add(zbozi3)

# V setu budou jen dvě unikátní položky (Mrkev, Celer), zbozi2 se neuloží
print("Počet unikátních položek v setu:", len(unikatni_zbozi))
```
