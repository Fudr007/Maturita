# 6. Asymptotické paměťové a časové složitosti

### Definice
* Asymptotická složitost vyjadřuje, jak se mění nároky algoritmu v závislosti na rostoucí velikosti vstupu

### Časová složitost
* Jak rychle roste čas potřebný k dokončení programu, když se zvětšuje počet vstupních dat,
* Lze ji měřit tak, že zaznamenáváme dobu běhu pro zvětšující se množiny prvků a výsledek zaneseme do grafu

### Paměťová složitost
* Jak roste množství alokované paměti RAM nezbytné pro průběh algoritmu
* Také jí lze měřit built-in nástroji

### Landauova notace (Big-O)
* K teoretickému zápisu složitosti
* Poskytuje zjednodušený pohled na limitní chování funkce, když se vstup blíží nekonečnu

### Závislost na typu algoritmu
* Algoritmy Brute force - prohledávají celý stavový prostor, čelíme faktoriálové časové složitosti $O(n!)$, extrémně rychle roste do neúnosných hodnot výpočetního času, ale je zaručeno že najde všechna řešení
* Heuristické algoritmy - využijí znalost problému, prohledávají jen jeho nezbytnou část díky nějakému "triku"
* Monte Carlo algoritmy - náhodné pokusy, limituje provedený čas, na úkor nalezení optimálního řešení, tzn. nemusí najít všechna řešení

```python
import time
import matplotlib.pyplot as plt
```

```python
def permutuj(uz_hotovo, zbyvajici_prvky, vysledek):
    # Rekurzivní funkce z materiálů pro generování permutací
    if not zbyvajici_prvky:
        vysledek.append(uz_hotovo)
    else:
        for i in range(len(zbyvajici_prvky)):
            permutuj(
                uz_hotovo + [zbyvajici_prvky[i]],
                zbyvajici_prvky[:i] + zbyvajici_prvky[i + 1:],
                vysledek
            )
```

**MĚŘENÍ ASYMPTOTICKÉ ČASOVÉ SLOŽITOSTI**

```python
velikosti_vstupu = [4, 5, 6, 7, 8, 9]
casy_behu = []

for n in velikosti_vstupu:
    pole = list(range(n))
    vysledek = []

    start = time.time()
    permutuj([], pole, vysledek)
    end = time.time()

    doba_behu = end - start
    casy_behu.append(doba_behu)
    print(f"Velikost n={n}, Počet permutací: {len(vysledek)}, Doba: {doba_behu:.5f} s")

# --- VYKRESLENÍ GRAFU ---
# Důkaz exponenciálně-faktoriálového růstu časové složitosti
plt.plot(velikosti_vstupu, casy_behu, marker='o', color='red', linewidth=2)
plt.title("Měření růstu časové složitosti algoritmu (Brute force)")
plt.xlabel("Velikost vstupního pole (n)")
plt.ylabel("Čas (sekundy)")
plt.grid(True)
plt.show()
```
