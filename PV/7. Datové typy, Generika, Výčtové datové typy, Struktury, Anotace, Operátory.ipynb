# 7. Datové typy, Generika, Výčtové datové typy, Struktury, Anotace, Operátory

### Datové typy
* Určují povahu proměnné
* Jaké operace s ní lze provádět a kolik paměti zabere
* Python je dynamicky typovaný, datový typ proměnné se může za běhu měnit
* Primitivní:
    * celá čísla (`int`)
    * desetinná čísla (`float`)
    * komplexní čísla (`complex`)
    * texty (`str`)
    * logické hodnoty (`bool`)
    * absence hodnoty (`None`)
* Kolekce:
    * seznam (`list`)
    * neměnný seznam (`tuple`)
    * množina (`set`)
    * slovník (`dict`)

### Operátory
* Symboly, které provádějí operace nad operandy
* Aritmetické:
    * sčítání (`+`)
    * odčítání (`-`)
    * násobení (`*`)
    * dělení (`/`)
    * modulo (`%`)
* Relační:
    * vrací `True`/`False` hodnoty
    * je rovno (`==`)
    * není rovno (`!=`)
    * menší než (`<`)
    * menší nebo rovno (`<=`)
    * větší než (`>`)
    * větší nebo rovno (`>=`)
* Logické:
    * `AND`
    * `OR`
    * `NOT`

### Enum
* Výčtový datový typ
* Může nabývat jen definovaných pevných hodnot
* Omezují chybovost a zvyšují čitelnost kódu

### Struktury (Struct)
* Způsob z jazyka C, jak seskupit související proměnné do jednoho celku pod jedním jménem
* Z velké části nahrazeny plnohodnotnými třídami
* V Pythonu se realizují přes tzv. `dataclasses`

### Generika
* Umožňuje psát kód tak, aby dokázal bezpečně pracovat s libovolným datovým typem
* Konkrétní typ se určí až když z generické třídy tvoříme instanci
* Výrazně redukují duplicity v kódu

### Anotace (Wrappery / Dekorátory)
* V Pythonu se značí `@` před definicí metody
* Návrhový vzor, kdy funkce na vstupu přijme jinou funkci a "obalí" ji dalšími příkazy
* Rozšíří její chování bez nutnosti modifikovat původní kód

```python
from enum import Enum
from dataclasses import dataclass
from typing import TypeVar, Generic
```

**1. ENUM**

```python
class StavZbozi(Enum):
    SKLADEM = 1
    VYPRODANO = 2
    NA_CESTE = 3

aktualni_stav = StavZbozi.SKLADEM
print("Je zboží skladem?", aktualni_stav == StavZbozi.SKLADEM)
```

**2. STRUKTURY**

```python
@dataclass
class Bod:
    x: int
    y: int

kotevni_bod = Bod(10, 20)
```

**3. GENERIKA**

```python
T = TypeVar('T')

class Zasilka(Generic[T]):
    def __init__(self, obsah: T):
        self.obsah = obsah

zasilka_s_cislem = Zasilka(42)
zasilka_s_textem = Zasilka("Dokument PDF")
```

**4. ANOTACE A WRAPPER**

```python
def wrapper_velka_pismena(puvodni_funkce):
    def obalena_funkce(*args, **kwargs):
        # Zavoláme původní funkci a získáme její výsledek
        vysledek = puvodni_funkce(*args, **kwargs)
        # Rozšíříme její chování
        return vysledek.upper()
    return obalena_funkce

# Použití anotace nad metodou
@wrapper_velka_pismena
def formatuj_cele_jmeno(jmeno, prijmeni):
    return f"{jmeno} {prijmeni}"

# Anotace zajistí, že i malá písmena na vstupu vyjdou jako velká
print("\nVýsledek anotované funkce:")
print(formatuj_cele_jmeno("jan", "novak"))
```
