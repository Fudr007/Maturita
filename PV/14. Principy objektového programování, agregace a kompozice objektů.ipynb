
# 14. Principy objektového programování, agregace a kompozice objektů


### Základní principy OOP
* Zapouzdření (Encapsulation)
    * Slouží ke skrytí vnitřní implementace objektu před okolním programem
    * V Pythonu nejsou klasické modifikátory (public, private), zapouzdření se realizuje pojmenovávajím s podtržítkem na začátku
    * Manipulace s nimi zvenčí probíhá výhradně přes přetížené metody
    * `@property` pro čtení a `@nazev.setter` pro zápis
* Dědičnost (Inheritance)
    * Vytváření nové třídy odvozením z jiné (rodičovské) třídy
    * V Pythonu zapis vložením názvu rodičovské třídy do kulatých závorek za název nové třídy (např. `class MojeException(Exception):`)
* Polymorfismus
    * Souvísí se schopností metod chovat se různě
    * Method overriding - Mění chování zděděné metody v odvozené třídě
    * Method overloading - Metody liší svou signaturou (vstupními parametry)
    * Volání virtuální metody následně závisí až na "runtime typu", tedy na konkrétním typu objektu za běhu programu

### Vztahy mezi objekty - Agregace a Kompozice
* V OOP se z menších objektů skládají větší celky
* Dva hlavní způsoby skládání jsou agregace a kompozice
* Agregace (Aggregation)
    * Volnější vztah celku a jeho částí
    * Umožňuje sdílení objektu
    * Část je vytvořena nezávisle a celku je předána pouze reference, pokud celek zanikne, vnitřní objekt v paměti zůstává a může ho používat jiný objekt
* Kompozice (Composition)
    * Těsnější vztah
    * Umožňuje měnit chování celku za běhu programu
    * Často se pojí s návrhovým vzorem Strategy - Objekt nevynucuje chování dědičností, ale deleguje ho na svou vnitřní komponentu, změnou komponenty za běhu se změní chování celého objektu

**1. PRINCIP ZAPOUZDŘENÍ**

```python
class Zbozi:
    def __init__(self):
        # Zapouzdření - skrytí implementace (proměnná začíná podtržítkem)
        self._vaha = 0

    @property
    def vaha(self):
        return self._vaha

    @vaha.setter
    def vaha(self, hodnota):
        if hodnota < 0:
            raise Exception("Záporná váha nesmí existovat")
        self._vaha = hodnota
```

**2. AGREGACE (Umožňuje sdílení objektu)**

```python
class Motor:
    def __init__(self, typ):
        self.typ = typ

class AutoAgregace:
    def __init__(self, motor: Motor):
        # Motor je vytvořen jinde a předán zvenku (sdílení reference)
        self.motor = motor

# Vytvoření nezávislého objektu Motor v hlavní paměti
sdileny_motor = Motor("1.9 TDI")
# Agregace: Předání reference umožňuje, aby dva nezávislé celky sdílely část
auto1 = AutoAgregace(sdileny_motor)
auto2 = AutoAgregace(sdileny_motor)
```

**3. KOMPOZICE (Umožňuje měnit chování za běhu)**

```python
class ZpusobPohybuBeh:
    def pohni_se(self):
        return "Postava běží rychle!"

class ZpusobPohybuChuzech:
    def pohni_se(self):
        return "Postava jde velmi pomalu."

class HerniPostava:
    def __init__(self):
        # Kompozice: Třída si vytvoří nebo je jí přiřazena vnitřní komponenta
        # určující chování. Toto chování není pevně dáno dědičností.
        self.komponenta_chovani = ZpusobPohybuChuzech()

    def zmen_chovani_za_behu(self):
        # Změna chování objektu za běhu programu výměnou komponenty
        self.komponenta_chovani = ZpusobPohybuBeh()

    def proved_akci(self):
        return self.komponenta_chovani.pohni_se()

# Ukázka kompozice v praxi
hrac = HerniPostava()
print("Původní stav:", hrac.proved_akci())
# Objekt změní své vnitřní fungování za běhu
hrac.zmen_chovani_za_behu()
print("Po změně chování:", hrac.proved_akci())
```
