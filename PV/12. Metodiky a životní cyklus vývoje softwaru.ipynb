# 12. Metodiky a životní cyklus vývoje softwaru

### Životní cyklus softwaru
* Popisuje standardizované fáze, každý software od úvodní myšlenky až po konec životnosti
* Analýza požadavků - Co přesně má systém dělat (sběr dat od zákazníka)
* Návrh - Tvorba systémové architektury, návrh databáze a uživatelského rozhraní
* Implementace - Programování zdrojového kódu, implementace návrhu
* Testování - Ověřování kvality, opravování nedostatků, hledání defektů a kontrola celé funkčnosti
* Nasazení - Instalace a předání softwaru do produkčního prostředí, reálné nasazení
* Údržba - Následné opravy chyb a přidávání funkcí na základě feedbacku od zákazníka a uživatelů

### Waterfall
* Tradiční sekvenční metodika
* Striktně lineární postup
* Další fáze životního cyklu začíná až ve chvíli, kdy je předchozí kompletně dokončena a schválena
* Výhody - Detailní počáteční dokumentace, snadné plánování termínů a finančního rozpočtu
* Nevýhody - Extrémní nepružnost. Pokud zákazník v průběhu implementace změní názor na nějakou funkci, je velmi těžké a drahé se vracet k analýze a designu. Funkční produkt je vidět až na konci cyklu

### Agilní metodiky
* Moderní iterativní přístup řešící nedostatky Waterfallu
* Vývoj neprobíhá lineárně v jednom velkém bloku, ale software vzniká postupně po menších přírůstcích
* Výhody - Průběžná komunikace se zákazníkem, schopnost okamžitě dělat změny požadavků i v pozdních fázích vývoje, zákazník může první funkce vyzkoušet brzy

### Scrum
* Konkrétní a nejpoužívanější rámec agilního vývoje
* Role:
    * Product Owner - Zastupuje zákazníka, spravuje seznam úkolů, určuje jejich prioritu
    * Scrum Master - Spravuje aby tým správně dodržoval agilní postupy, odstraňuje překážky brzdící vývoj
    * Development Team - Křížově funkční tým programátorů a testerů, který práci reálně vykonává
* Události:
    * Sprint - Časově ohraničená vývojová iterace (obvykle 2 až 4 týdny max), na konci musí být funkční část softwaru
    * Stand-up - Každodenní 15minutová synchronizace týmu ("Co jsem dělal, co budu dělat, co mě blokuje")
    * Sprint Planning - Plánování, co přesně se stihne vyrobit v následující iteraci Sprintu
    * Retrospective - Ohlédnutí týmu na konci Sprintu s cílem zlepšit svůj proces do budoucna
* Artefakty:
    * Product Backlog - Dynamický seznam všech požadavků na výsledný produkt
    * Sprint Backlog - Vybrané úkoly z Product Backlogu, na kterých se pracuje v probíhajícím Sprintu

**Test-Driven Development**

```python
import unittest

#Testy pro app
class TestApp(unittest.TestCase):
    def test_vypocet_slevy(self):
        vysledek = aplikuj_slevu(1000, 20)
        self.assertEqual(vysledek, 800)

    def test_neplatna_sleva(self):
        with self.assertRaises(ValueError):
            aplikuj_slevu(1000, 150)

if __name__ == '__main__':
    #Implementace funkce podle testu
    def aplikuj_slevu(cena:float, sleva):
        if sleva > 100 or sleva <= 0:
            raise ValueError("Sleva musi byt v rozmezi 1 a 100")
        return cena * (1 - sleva / 100)

    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
